```python
from youtube_transcript_api import YouTubeTranscriptApi, TranscriptsDisabled
```

1. Indexing

a. Document Ingestion

```python
video_id = "LPZh9BOjkQs" # only id and not full URL
```

```python
try : 
    yt_api = YouTubeTranscriptApi()
    fetched_transcript = yt_api.fetch(video_id, languages = ["en"])  # .fetch() as an instance method so it is mandatory to use it using an object.
    transcript = ' '.join(i.text for i in fetched_transcript)
    print(transcript)

except TranscriptsDisabled: 
    print("No Caption available for this video.")
```

1. Indexing

b. Text Splitting

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
splitter = RecursiveCharacterTextSplitter(chunk_size = 1000, chunk_overlap = 200)
```

```python
docs = splitter.create_documents([transcript])  # create_documents() takes list as input
print(docs)
```

```python
print(len(docs))
```

```python
docs = docs[1:]
```

```python
print(len(docs))
```

1. Indexing

b. Embedding

c. Storing in Vector Stores

```python
from langchain_google_genai import GoogleGenerativeAIEmbeddings
from dotenv import load_dotenv
from langchain_community.vectorstores import FAISS
```

```python
load_dotenv()
```

```python
embedding_model = GoogleGenerativeAIEmbeddings(model = "models/gemini-embedding-001")
```

```python
vector_store = FAISS.from_documents(docs, embedding_model)
```

```python
print(vector_store)
```

```python
vector_store.index_to_docstore_id
```

```python
vector_store.get_by_ids(['2554ab4f-9760-4169-8dbb-b0a40c2e7d37'])
```

```python
vector_store.get_by_ids(['2554ab4f-9760-4169-8dbb-b0a40c2e7d37'])[0].page_content
```

2. Retreival

```python
retriever = vector_store.as_retriever(search_type = "similarity", search_kwargs={"k": 3})
```

```python
retriever.invoke("what is LLM?")
```

3. Augmentation


```python
from langchain_core.prompts import ChatPromptTemplate
```

```python
prompt = ChatPromptTemplate.from_template("""You're a helpful assistant. Answer only from the provided context. 
                                          If the context is insufficient, simply respond you don't know. 
                                          Context : {context} Question : {question}""")
```

```python
user_query = "Are transformers discussed? If yes, then answer in 2 lines about them."
retrieved_docs = retriever.invoke(user_query)
```

```python
print(retrieved_docs)
```

```python
print(retrieved_docs[0].page_content)
```

```python
# context will be the concatenation of all retrieved docs
context = "\n\n".join(doc.page_content for doc in retrieved_docs)
print(context)
```

```python
final_prompt = prompt.invoke({"context" : context, "question" : user_query})
print(final_prompt)
```

4. Generation

```python
from langchain_google_genai import ChatGoogleGenerativeAI
```

```python
llm = ChatGoogleGenerativeAI(temperature = 0.2, model = "gemini-2.5-flash")
response = llm.invoke(final_prompt)
```

```python
print(response)
```

```python
print(response.content)
```

```python
from langchain_core.runnables import RunnableParallel, RunnableLambda, RunnablePassthrough
```

```python
def final_context(retrieved_docs):
    context = "\n\n".join(doc.page_content for doc in retrieved_docs)
    return context
```

```python
primary_chain = RunnableParallel(
    {
        "question" : RunnablePassthrough(),
        "context" : retriever | RunnableLambda(final_context)
    }
)
```

```python
primary_chain.invoke("what are Transformers?")
```

```python
from langchain_core.output_parsers import StrOutputParser
```

```python
parser = StrOutputParser()
```

```python
main_chain = primary_chain | prompt | llm | parser
```

```python
answer = main_chain.invoke("what are aliens?")
```

```python
print(answer)
```
